## Mean
Mean = The average value of a set of numbers it represents the central value

### Formula
Mean = (Sum of all values) / (Number of values)

Mean (x̄) = Σx / n

### Numerical Example
Data: 99, 87, 0, 65, 62
Mean = (99+87+0+65+62 / 5 = 314 / 5 = 62.6

### Business Example
Imagine you are a cricket coach and you have many players. After every 5 or 10 matches, you want to calculate the average score of each player. For that, you can use Mean.
If Player BK scored [99, 87, 0, 65, 62], we add all the scores and divide the total by the number of matches.
Mean = (99 + 87 + 0 + 65 + 62) ÷ 5 = 62.6
So, Player BK's average score is 62.6 runs.

### AI/ML Example
Mean is used to fill missing values in a dataset (mean imputation), 
and it's the basis for many loss functions in ML, like Mean 
Squared Error (MSE), which measures average prediction error.

### Interpretation
Player BK scored [99, 87, 0, 65, 62], giving a mean of 62.6 runs. This means that if all 5 match scores were equal, each match would have a score of 62.6 runs, so 62.6 represents the balance point of the player's scores. However, the mean is sensitive to extreme values or outliers. For example, if one match had a score of 200 instead of 62, the mean would increase significantly because that unusually high score pulls the mean upward.

```python
import numpy as BK
data = [99, 87, 0, 65, 62]
Average_score = BK.mean(data)
print("Mean:", Average_score)
```

```python
import numpy as Store
data = [20000, 30000, 7888, 43984]
Store_sales = Store.mean(data)
print("Mean:", Store_sales)
```

## 2. Median
### Definition
Median is the middle value of the data set after arranging it into order
If there's an even number of values, it's the average of the two middle values


### Formula
If n (number of values) is odd:
Median = value at position (n+1)/2, after sorting

If n is even:
Median = average of values at positions n/2 and (n/2 + 1), after sorting

### Numerical Example
Data (odd count): 10, 20, 30, 40, 50 → sorted already → Median = 30 (middle value)
Data (even count): 10, 20, 30, 40 → Median = (20+30)/2 = 25

Data (even count): 10, 20, 30, 40 → Median = (20+30)/2 = 25

### Business Example
A company reports employee salaries: ₹25000, ₹28000, ₹30000, 
₹32000, ₹500000 (one CEO). The MEDIAN salary gives a more honest 
"typical employee" picture than the mean, since the CEO's huge 
salary would badly skew the mean upward.

### AI/ML Example
Median is often used instead of mean for filling missing values 
when a feature has outliers or is skewed - like house prices or 
income data, where a few extreme values would distort a mean-based fill.

### Interpretation
Median tells you the value that splits the data exactly in 
half - 50% of values are below it, 50% are above it. Unlike mean, 
it's NOT affected by extreme outliers, making it more "robust" for 
skewed data.

```python
import numpy as np

data_odd = [10, 20, 30, 40, 50]
data_even = [10, 20, 30, 40]

print("Median (odd count):", np.median(data_odd))
print("Median (even count):", np.median(data_even))

# Demonstrating median's robustness to outliers, vs mean
salaries = [25000, 28000, 30000, 32000, 500000]   # one extreme outlier (CEO)
print("\nMean salary:", np.mean(salaries))       # heavily pulled up by outlier
print("Median salary:", np.median(salaries))     # stays representative

# Where it's used: np.median() finds the true middle value after
# sorting internally. Notice how the mean salary (~123,000) is
# badly distorted by the one CEO outlier, while the median (30,000)
# still reflects what a TYPICAL employee actually earns - this is
# the core reason median is preferred for skewed real-world data.
```

## 3. Mode

### Definition
The Mode is the value that appears MOST FREQUENTLY in a dataset. 
Unlike mean and median, a dataset can have one mode, multiple 
modes, or no mode at all (if every value appears equally often).

### Formula
No mathematical formula - Mode is found by counting frequency of 
each value and identifying the most common one.

### Numerical Example
[50, 70, 50, 90, 50, 80, 70]
50 → 3 times
70 → 2 times
90 → 1 time
80 → 1 time
 so mode = 50

### Business Example
A shoe store tracks which shoe SIZE sells most often: sizes 7, 8, 
8, 9, 8, 10. The Mode (size 8) tells the store which size to stock 
the MOST of - more useful here than a mean size, which wouldn't 
correspond to a real, orderable shoe size.

### AI/ML Example
Mode is commonly used to fill missing values for CATEGORICAL 
features (like "most common city" or "most common product 
category"), since mean/median don't make sense for non-numeric data.

### Interpretation
A mode of 20 means 20 is the single most "popular" or frequently 
occurring value in the dataset. It's the only measure of central 
tendency that works meaningfully for categorical (qualitative) 
data, not just numbers.



```python
from scipy import stats
import numpy as np

data = [10, 20, 20, 30, 40, 20, 50]
mode_result = stats.mode(data, keepdims=True)
print("Mode:", mode_result.mode[0], "| Count:", mode_result.count[0])

# Mode also works on categorical data - shoe sizes example
shoe_sizes = [7, 8, 8, 9, 8, 10]
mode_shoe = stats.mode(shoe_sizes, keepdims=True)
print("Most common shoe size:", mode_shoe.mode[0])

# Where it's used: stats.mode() counts how often each value appears
# and returns the most frequent one. In the first dataset, 20
# appears 3 times (more than any other value), confirming it's the
# mode. Same logic applies to shoe_sizes - size 8 appears most often.
```

## 4. Weighted Mean

### Definition
Weighted Mean = Average where each value has a different level of importance. based on an assigned "weight" or importance - unlike 
a regular mean, which treats every value equally.

### Formula
Weighted Mean = Σ(value × weight) / Σ(weight)

### Numerical Example
Exam scores with different weightages: 
Assignment = 80 (weight 20%), Midterm = 70 (weight 30%), 
Final = 90 (weight 50%)

Weighted Mean = (80×0.20 + 70×0.30 + 90×0.50) / (0.20+0.30+0.50)
             = (16 + 21 + 45) / 1.0 = 82

### Business Example
A product's overall rating combines reviews from different 
platforms, but platform A (5000 reviews) should count more than 
platform B (50 reviews) - a weighted mean reflects true customer 
sentiment better than treating both platforms equally.

### AI/ML Example
Weighted Mean is used in ensemble ML models, where predictions 
from multiple models are combined - a more accurate model might be 
given a higher WEIGHT than a less accurate one when averaging their predictions.

### Interpretation
A weighted mean of 82 (vs a plain average of 80) shows that giving 
the Final exam more importance (50% weight) pulled the overall 
score up, since the student scored highest on the Final.


```python
import numpy as np

scores = [80, 70, 90]
weights = [0.20, 0.30, 0.50]

weighted_mean = np.average(scores, weights=weights)
plain_mean = np.mean(scores)

print("Plain Mean:", plain_mean)
print("Weighted Mean:", weighted_mean)

# Where it's used: np.average(scores, weights=weights) multiplies
# each score by its weight before averaging - notice the weighted
# mean (82) differs from the plain mean (80), since the Final exam
# (90, highest score) was given the most weight (50%), pulling the
# overall result upward.
```

## 5. Geometric Mean

Geometric Mean = An average calculated by multiplying all the values and then taking the nth root

 used 
specifically for data that grows multiplicatively, like growth 
rates or percentages, rather than data you'd simply add together.

### Formula
Geometric Mean = (x₁ × x₂ × x₃ × ... × xₙ)^(1/n)

### Numerical Example
Data (growth rates as multipliers): 1.10, 1.20, 1.05 (10%, 20%, 5% growth)
Geometric Mean = (1.10 × 1.20 × 1.05)^(1/3) = (1.386)^(1/3) ≈ 1.1147
→ approximately 11.47% average growth rate per period

### Business Example
A company's revenue grows 10% in year 1, 20% in year 2, and 5% in 
year 3. The Geometric Mean gives the TRUE average annual growth 
rate - a plain (arithmetic) mean would overstate it, since growth 
compounds rather than adds.

### AI/ML Example
Geometric Mean is used in evaluating model performance across 
multiple metrics that have very different scales (e.g., combining 
precision and recall scores fairly, since arithmetic mean can be 
misleading when one metric is much larger than another).

### Interpretation
A geometric mean growth rate of ~11.47% means if the company grew 
at exactly this SAME rate every year, it would end up at the same 
final value as the actual varying growth rates did - it reflects 
compounding, unlike a plain average.


```python
from scipy import stats

growth_rates = [1.10, 1.20, 1.05]   # representing 10%, 20%, 5% growth
geometric_mean = stats.gmean(growth_rates)
print("Geometric Mean growth rate:", geometric_mean)
print("As percentage growth:", (geometric_mean - 1) * 100, "%")

# Where it's used: stats.gmean() multiplies all three growth
# multipliers together, then takes the cube root (since there are 3
# values) - giving the TRUE compounding average growth rate, which
# differs from a plain arithmetic mean of the same numbers.
```

## 6. Harmonic Mean

### Definition
The Harmonic Mean is the reciprocal of the average of reciprocals 
of the values - used specifically for RATES (like speed, or 
ratios), where averaging directly would give a misleading result.

### Formula
Harmonic Mean = n / Σ(1/x)

### Numerical Example
A car travels the same distance at 60 km/h going, and 40 km/h 
returning.
Harmonic Mean = 2 / (1/60 + 1/40) = 2 / (0.0167 + 0.025) = 2 / 0.0417 ≈ 48 km/h
(NOT the plain average of 50 km/h - that would be wrong for this scenario)

### Business Example
If a delivery company's vehicles average 60 km/h on one leg and 40 
km/h on the return leg of the SAME route, the harmonic mean gives 
the TRUE average speed for the round trip - a plain average would 
overstate actual performance.

### AI/ML Example
Harmonic Mean is the basis of the F1-Score in ML (combining 
Precision and Recall) - it's used specifically because it 
penalizes imbalance between the two metrics more than a plain 
average would, giving a fairer combined score.

### Interpretation
A harmonic mean speed of 48 km/h (not 50) reflects that the car 
spent MORE TIME traveling at the slower speed (40 km/h) than the 
faster one, since time = distance/speed - harmonic mean correctly 
accounts for this, unlike a plain average.

```python
from scipy import stats

speeds = [60, 40]   # km/h, going and returning over the same distance
harmonic_mean = stats.hmean(speeds)
print("Harmonic Mean speed:", harmonic_mean, "km/h")
print("Compare to plain average:", sum(speeds)/len(speeds), "km/h (misleading here)")

# Where it's used: stats.hmean() correctly accounts for the fact
# that the car spends MORE TIME at the slower speed - the harmonic
# mean (48 km/h) is lower than the plain average (50 km/h), which
# reflects the true average speed for the round trip more accurately.
```
