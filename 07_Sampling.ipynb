## 1. Sampling

### Explanation
Sampling is the process of selecting a SUBSET of individuals from 
a larger population to study, instead of collecting data from 
every single member - used because studying an entire population 
is often too costly, slow, or impossible.

### Real-World Analogy
A chef tastes one spoonful of soup (a sample) to judge the flavor 
of the entire pot (population), instead of drinking the whole pot.

### AI/ML Use Case
Training data for any ML model is always a SAMPLE of all possible 
real-world inputs the model will encounter - the quality of that 
sample directly affects how well the model generalizes.

```python
import numpy as np

population = np.arange(1, 1001)   # imagine 1000 total people/items
sample = np.random.choice(population, size=50, replace=False)

print("Population size:", len(population))
print("Sample size:", len(sample))
print("Sample mean:", np.mean(sample), "| Population mean:", np.mean(population))

# Where it's used: population represents ALL 1000 members.
# np.random.choice(..., size=50) selects just 50 of them - the
# SAMPLE. Notice the sample mean is close to, but not exactly, the
# true population mean - a normal and expected result of sampling.
```

## 2. Random Sampling

### Explanation
Random Sampling means every member of the population has an EQUAL 
chance of being selected - no bias toward any particular group or 
individual, pure chance decides who's included.

### Real-World Analogy
Pulling names out of a hat, fully mixed, to choose winners for a 
raffle - every name has the same chance of being picked.

### AI/ML Use Case
Random Sampling is used when splitting a dataset into training and 
test sets (train_test_split) - ensuring the split doesn't 
accidentally favor any particular pattern or order in the original data.

```python
import numpy as np

population = np.arange(1, 101)   # 100 people, numbered 1-100
random_sample = np.random.choice(population, size=10, replace=False)

print("Random Sample:", sorted(random_sample))

# Where it's used: np.random.choice(..., replace=False) picks 10
# values from the population with NO systematic pattern - every
# number from 1-100 had an equal chance of being selected, the
# defining trait of random sampling.
```

## 3. Stratified Sampling

### Explanation
Stratified Sampling divides the population into distinct SUBGROUPS 
(strata) based on a shared characteristic (like age group or 
gender), then randomly samples from EACH subgroup proportionally - 
ensuring every subgroup is properly represented in the final sample.

### Real-World Analogy
A survey wants to represent all age groups fairly. Instead of 
randomly picking from everyone (which might accidentally get mostly 
young people), it splits people into age groups (18-30, 31-50, 
51+) and randomly samples from EACH group separately, in 
proportion to their size in the population.

### AI/ML Use Case
Stratified Sampling is used when splitting imbalanced datasets 
(e.g., a rare disease dataset with 95% healthy, 5% sick) into 
train/test sets, ensuring both sets maintain the SAME class 
proportion - preventing a test set that accidentally has zero sick 
patients.

```python
import pandas as pd

data = pd.DataFrame({
    "student_id": range(1, 21),
    "course": ["AI/ML"]*12 + ["Web Dev"]*8
})

# Simpler and version-safe: groupby().sample() does stratified sampling directly
stratified_sample = data.groupby("course", group_keys=False).sample(frac=0.5, random_state=1)

print(stratified_sample["course"].value_counts())
```

## 4. Cluster Sampling

### Explanation
Cluster Sampling divides the population into naturally occurring 
GROUPS (clusters), then randomly selects ENTIRE clusters to study - 
rather than sampling individuals from within every group like 
Stratified Sampling does.

### Real-World Analogy
A researcher wants to survey students across a country. Instead of 
randomly picking individual students from every school (expensive, 
slow), they randomly pick a few ENTIRE SCHOOLS (clusters) and 
survey everyone within those chosen schools.

### AI/ML Use Case
Used in large-scale data collection where collecting from every 
individual is impractical - like selecting a few entire IoT sensor 
regions (clusters) to analyze in depth, 
rather than sampling 
individual sensors scattered everywhere.

```python
import numpy as np

# 5 clusters (schools), each with 10 students
clusters = {f"School_{i}": np.arange(i*10, i*10+10) for i in range(1, 6)}

# Randomly select 2 ENTIRE clusters (schools)
np.random.seed(1)
selected_clusters = np.random.choice(list(clusters.keys()), size=2, replace=False)

print("Selected clusters (entire schools sampled):", selected_clusters)
for school in selected_clusters:
    print(f"{school} students:", clusters[school])

# Where it's used: instead of picking individual students from ALL
# 5 schools, np.random.choice selects just 2 WHOLE schools - every
# student within those 2 chosen schools is included, while the
# other 3 schools are entirely excluded. This is the key difference
# from Stratified Sampling, which samples FROM every group.
```

## 5. Systematic Sampling

### Explanation
Systematic Sampling selects every Nth member from a population 
after a random starting point - creating an evenly-spaced sample, 
rather than a purely random one.

### Real-World Analogy
A quality inspector checks every 10th item coming off a factory 
conveyor belt, rather than randomly picking items - simple to 
implement and still gives good coverage across the whole production run.

### AI/ML Use Case
Used when collecting time-series data at regular intervals (e.g., 
sensor readings every 5 minutes) - the systematic interval itself 
becomes a natural sampling method for building a representative dataset.

```python
import numpy as np

population = np.arange(1, 101)   # 100 items
k = 10   # sample every 10th item
start = np.random.randint(0, k)   # random starting point

systematic_sample = population[start::k]
print("Systematic Sample (every 10th item):", systematic_sample)

# Where it's used: population[start::k] picks one random starting
# point, then selects every 10th value from there - creating an
# evenly-spaced sample across the whole population, unlike the
# purely random selection we saw in Random Sampling.
```

## 6. Sampling Bias

### Explanation
Sampling Bias occurs when the sample is NOT representative of the 
population, because some members were systematically more likely 
to be included than others - leading to skewed, misleading conclusions.

### Real-World Analogy
A phone survey conducted only during work hours will oversample 
retirees and unemployed people, while undersampling working 
professionals who can't answer calls then - the sample doesn't 
truly reflect the whole population.

### AI/ML Use Case
If training data for a facial recognition model comes mostly from 
one demographic group, the model will perform poorly on 
underrepresented groups - a well-documented real-world ML fairness 
problem caused directly by sampling bias.

```python
import numpy as np

population_ages = np.random.randint(18, 80, size=1000)   # true population: ages 18-80

# Biased sample - only sampling from a narrow, non-representative slice
biased_sample = population_ages[population_ages < 30]   # only young people, systematically excluding others

print("True population mean age:", np.mean(population_ages))
print("Biased sample mean age:", np.mean(biased_sample))
print("Notice the large gap - this sample badly misrepresents the true population.")

# Where it's used: filtering population_ages < 30 deliberately
# creates a BIASED sample - it systematically excludes everyone 30+.
# The resulting sample mean is far lower than the true population
# mean, demonstrating exactly how sampling bias distorts conclusions.
```

## 7. Sampling Error

### Explanation
Sampling Error is the natural, unavoidable difference between a 
sample statistic (like sample mean) and the true population 
parameter (population mean) - it exists even with a PERFECTLY 
random, unbiased sample, simply because a sample is not the entire 
population.

### Real-World Analogy
Even a perfectly fair, random poll of 1000 voters won't predict an 
election with 100% precision - there's always some natural "wiggle 
room" (margin of error) simply because not every voter was asked, 
even if there was zero bias in who was chosen.

### AI/ML Use Case
Sampling error is why model performance metrics (like accuracy) 
can vary slightly between different random train/test splits of 
the SAME dataset - part of the observed variation is genuine 
sampling error, not necessarily a real difference in the model.

```python
import numpy as np

population = np.random.normal(loc=50, scale=10, size=10000)
true_population_mean = np.mean(population)

# Taking multiple random samples and observing the error each time
for i in range(5):
    sample = np.random.choice(population, size=30, replace=False)
    sampling_error = np.mean(sample) - true_population_mean
    print(f"Sample {i+1} mean: {np.mean(sample):.2f} | Sampling Error: {sampling_error:.2f}")

# Where it's used: each of the 5 random samples gives a SLIGHTLY
# different mean, even though all are drawn correctly and randomly
# from the SAME population. That natural variation each time is
# sampling error - unavoidable, but usually small with good sample sizes.
```

## 8. Sample Size

### Explanation
Sample Size is the number of individuals/observations included in 
a sample. Larger sample sizes generally lead to more accurate 
estimates (less sampling error), but with diminishing returns - 
doubling sample size doesn't halve the error, since error 
decreases with the SQUARE ROOT of sample size.

### Formula
Standard Error ≈ (Population Standard Deviation) / √(Sample Size)

### Real-World Analogy
Tasting one spoonful of soup gives a rough idea of flavor; tasting 
five spoonfuls from different parts of the pot gives a much more 
reliable read - but tasting 50 spoonfuls doesn't make you 50x more 
confident, just somewhat more confident, with diminishing benefit.

### AI/ML Use Case
Determining adequate sample size is critical before collecting 
training data - too small a sample leads to a model that doesn't 
generalize well, while collecting far more data than needed wastes 
time/cost for only marginal accuracy gains.

```python
import numpy as np

population = np.random.normal(loc=50, scale=10, size=100000)
true_mean = np.mean(population)

sample_sizes = [10, 50, 200, 1000]
for size in sample_sizes:
    sample = np.random.choice(population, size=size, replace=False)
    error = abs(np.mean(sample) - true_mean)
    print(f"Sample size {size}: Error from true mean = {error:.3f}")

# Where it's used: as sample_sizes increases (10 -> 1000), the
# error generally shrinks - but notice it doesn't shrink in a
# straight line proportional to size, since error scales with the
# SQUARE ROOT of sample size, exactly as the formula describes.
```
