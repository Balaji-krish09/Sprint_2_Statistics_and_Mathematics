## 1. Histogram
**Purpose:** Shows the distribution/frequency of a single numeric variable.
**When to use:** Understanding the shape, spread, and skewness of one variable.
**Advantages:** Simple, reveals distribution shape at a glance.
**Limitations:** Bin size choice can distort the visual interpretation.

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.normal(50, 10, 1000)
plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Histogram")
plt.show()
# Interpretation: shows the bell-shaped spread of the data around the mean.
```

## 2. Bar Chart
**Purpose:** Compares values across discrete CATEGORIES.
**When to use:** Comparing a numeric value (like sales, count) across categorical groups.
**Advantages:** Easy to read and compare categories directly.
**Limitations:** Not suitable for continuous numeric data.

```python
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 15]

plt.bar(categories, values, color="coral")
plt.title("Bar Chart")
plt.show()
# Interpretation: Category B has the highest value, D the lowest.
```

## 3. Line Chart
**Purpose:** Shows trends/changes over a continuous sequence, usually time.
**When to use:** Time-series data - tracking how a value changes over time.
**Advantages:** Clearly shows trends, patterns, and rate of change.
**Limitations:** Not suitable for unordered categorical data.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
sales = [200, 220, 210, 250, 270, 260, 300]

plt.plot(days, sales, marker="o", color="green")
plt.title("Line Chart - Sales Over Time")
plt.show()
# Interpretation: sales show a general upward trend across the week.
```

## 4. Scatter Plot (recap)
**Purpose:** Shows the relationship between TWO numeric variables.
**When to use:** Checking correlation or patterns between two continuous variables.
**Advantages:** Reveals correlation direction/strength and outliers.
**Limitations:** Hard to interpret with too many overlapping points (overplotting).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(50, 10, 100)
y = x * 2 + np.random.normal(0, 10, 100)

plt.scatter(x, y, alpha=0.6, color="purple")
plt.title("Scatter Plot")
plt.show()
# Interpretation: a clear upward trend suggests positive correlation between x and y.
```

## 5. Pie Chart
**Purpose:** Shows PROPORTIONS of a whole, as percentage slices.
**When to use:** Showing composition/share of categories out of a total (few categories only).
**Advantages:** Intuitive for showing simple proportions.
**Limitations:** Hard to compare slices accurately with many categories or similar sizes.

```python
import matplotlib.pyplot as plt

labels = ["Product A", "Product B", "Product C"]
sizes = [45, 30, 25]

plt.pie(sizes, labels=labels, autopct="%1.1f%%", colors=["gold", "lightblue", "lightgreen"])
plt.title("Pie Chart")
plt.show()
# Interpretation: Product A holds the largest share of sales at 45%.
```

## 6. Density Plot
**Purpose:** Shows a SMOOTHED version of a distribution (like a histogram, but continuous).
**When to use:** Understanding the shape/spread of continuous data, especially comparing multiple distributions.
**Advantages:** Smoother than histogram, easier to compare multiple groups.
**Limitations:** Can obscure real data details behind the smoothing.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

data = np.random.normal(50, 10, 1000)
density = gaussian_kde(data)
x_vals = np.linspace(min(data), max(data), 200)

plt.plot(x_vals, density(x_vals), color="darkblue")
plt.fill_between(x_vals, density(x_vals), alpha=0.3)
plt.title("Density Plot")
plt.show()
# Interpretation: smooth bell-shaped curve confirms the underlying normal distribution.
```

## 7. Box Plot (recap)
**Purpose:** Summarizes distribution using 5-number summary and shows outliers.
**When to use:** Comparing spread/outliers across one or more groups.
**Advantages:** Compact, great for comparing multiple groups side by side.
**Limitations:** Doesn't show the full distribution shape (e.g., can't tell bimodal data apart).

```python
import matplotlib.pyplot as plt

group_a = [20, 22, 21, 23, 19, 50]
group_b = [30, 32, 31, 33, 29, 34]

plt.boxplot([group_a, group_b], labels=["Group A", "Group B"])
plt.title("Box Plot")
plt.show()
# Interpretation: Group A has an outlier (50) and lower median than Group B.
```

## 8. Violin Plot
**Purpose:** Combines a Box Plot with a Density Plot - shows both summary 
statistics AND the full distribution shape.
**When to use:** When you want distribution shape details AND summary 
stats together, especially for comparing groups.
**Advantages:** More informative than a box plot alone (shows multimodal shapes).
**Limitations:** Harder to read for beginners; less compact than a box plot.

```python
import matplotlib.pyplot as plt
import numpy as np

group_a = np.random.normal(50, 5, 200)
group_b = np.random.normal(60, 10, 200)

plt.violinplot([group_a, group_b])
plt.xticks([1, 2], ["Group A", "Group B"])
plt.title("Violin Plot")
plt.show()
# Interpretation: Group B is wider (more spread/variance) than Group A, 
# visible in the shape's width, not just a box plot's whiskers.
```

## 9. Pair Plot
**Purpose:** Shows scatter plots for EVERY pair of numeric variables in a 
dataset at once, in a grid.
**When to use:** Quickly exploring relationships between MANY variables 
during initial data exploration (EDA).
**Advantages:** Comprehensive overview of all pairwise relationships in one view.
**Limitations:** Becomes cluttered/slow with many variables (10+ features).

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "age": [25, 32, 47, 51, 62, 28, 40],
    "income": [30000, 45000, 60000, 65000, 70000, 35000, 55000],
    "spending": [80, 70, 50, 45, 30, 75, 55]
})

pd.plotting.scatter_matrix(data, figsize=(8, 8), color="teal")
plt.suptitle("Pair Plot")
plt.show()
# Interpretation: the grid shows every variable plotted against every 
# other, letting us spot patterns (like age vs spending) all at once.
```

## 10. Heatmap (recap)
**Purpose:** Color-codes a matrix of values (commonly correlation) for 
instant visual pattern recognition.
**When to use:** Visualizing correlation matrices or any 2D grid of values.
**Advantages:** Instantly highlights strong relationships/patterns via color.
**Limitations:** Color perception can be subjective; needs a clear legend/scale.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "age": [25, 32, 47, 51, 62],
    "income": [30000, 45000, 60000, 65000, 70000],
    "spending": [80, 70, 50, 45, 30]
})

corr = data.corr()
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Heatmap")
plt.show()
# Interpretation: dark red/blue cells indicate strong positive/negative 
# correlation; here age-income are positively correlated, both negatively with spending.
```
