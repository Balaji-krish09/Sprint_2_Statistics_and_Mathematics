## 2. Multicollinearity

### Definition
Multicollinearity occurs when two or more FEATURES in a dataset 
are highly correlated with EACH OTHER (not just with the target) - 
causing problems for models like Linear Regression, which assume 
features are independent.

### Real-World Example
"Height in cm" and "Height in inches" as two separate features are 
perfectly multicollinear - they carry the exact same information, 
just in different units.

### AI/ML Example
Multicollinearity makes it hard to interpret individual feature 
coefficients in Linear Regression - the model can't clearly tell 
which of two highly correlated features is "really" driving the prediction.

```python
import pandas as pd

data = pd.DataFrame({
    "height_cm": [160, 165, 170, 175, 180],
    "height_inches": [63.0, 65.0, 66.9, 68.9, 70.9],   # same info, different unit
    "weight": [55, 60, 68, 72, 80]
})
print(data.corr())

# Where it's used: height_cm and height_inches show correlation
# very close to 1.0 - a textbook case of multicollinearity, since
# they're mathematically almost the same variable in different units.
```

## 3. Variance Inflation Factor (VIF)

### Definition
VIF quantifies HOW MUCH a feature's variance is inflated due to 
multicollinearity with other features. A VIF above 5-10 is 
generally considered a warning sign of problematic multicollinearity.

### Formula
VIF = 1 / (1 - R²)
Where R² comes from regressing that feature against all other features.

### AI/ML Example
Before finalizing features for a Linear Regression model, engineers 
calculate VIF for each feature - dropping or combining features 
with high VIF to improve model stability and interpretability.

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

data = pd.DataFrame({
    "height_cm": [160, 165, 170, 175, 180],
    "height_inches": [63.0, 65.0, 66.9, 68.9, 70.9],
    "weight": [55, 60, 68, 72, 80]
})

vif_data = pd.DataFrame()
vif_data["feature"] = data.columns
vif_data["VIF"] = [variance_inflation_factor(data.values, i) for i in range(data.shape[1])]
print(vif_data)

# Where it's used: variance_inflation_factor() calculates VIF for
# each column - height_cm and height_inches will show very HIGH VIF
# values, numerically confirming the multicollinearity we saw in
# the correlation matrix above.
```

## 4 & 5. Feature Importance and Feature Selection Basics

### Explanation
Feature Importance ranks how much each feature contributes to a 
model's predictions (commonly derived from tree-based models like 
Random Forest). Feature Selection is the broader process of 
choosing which features to KEEP for training, based on importance, 
correlation, or domain knowledge - improving model performance and reducing complexity.

### AI/ML Example
A Random Forest model trained on customer data might reveal that 
"income" and "past purchase count" are highly important features, 
while "favorite color" contributes almost nothing - guiding which 
features to keep vs drop for a simpler, faster model.

```python
from sklearn.ensemble import RandomForestRegressor
import pandas as pd

data = pd.DataFrame({
    "age": [25, 32, 47, 51, 62, 28, 40],
    "income": [30000, 45000, 60000, 65000, 70000, 35000, 55000],
    "favorite_color_code": [1, 3, 2, 1, 3, 2, 1],   # unlikely to matter
})
target = [80, 70, 50, 45, 30, 75, 55]   # spending score

model = RandomForestRegressor(random_state=1)
model.fit(data, target)

importance = pd.Series(model.feature_importances_, index=data.columns)
print(importance.sort_values(ascending=False))

# Where it's used: model.feature_importances_ ranks each feature's
# contribution to predictions - income and age should rank far
# higher than favorite_color_code, confirming what intuition
# already suggests: irrelevant features contribute little to the model.
```
