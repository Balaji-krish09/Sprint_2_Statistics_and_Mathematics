## 1. Null Hypothesis (H0) and Alternative Hypothesis (H1)

### Explanation
The Null Hypothesis (H0) assumes NO effect or NO difference exists - 
it's the default "nothing special is happening" claim we try to 
disprove. The Alternative Hypothesis (H1) is what we're actually 
trying to find evidence FOR - that there IS a real effect or difference.

### Business Example
Testing a new website design: H0 = "the new design has no effect 
on conversion rate." H1 = "the new design DOES change the 
conversion rate." We collect data to see if we can reject H0.

### AI/ML Example
When comparing two models' accuracy, H0 = "there's no real 
difference between Model A and Model B's performance." Hypothesis 
testing helps determine if an observed accuracy difference is 
statistically meaningful or just random chance.

```python
import numpy as np
from scipy import stats

# Website conversion rates: old design vs new design (sample data)
old_design = [0.10, 0.12, 0.11, 0.09, 0.10]
new_design = [0.15, 0.14, 0.16, 0.13, 0.15]

t_stat, p_value = stats.ttest_ind(old_design, new_design)
print("H0: No difference between old and new design")
print("T-statistic:", t_stat, "| p-value:", p_value)

# Where it's used: this test checks whether the observed difference
# in conversion rates is statistically significant, or could
# reasonably happen by random chance under H0 - the p-value below
# will tell us which, covered next.
```

## 2. p-value and Significance Level

### Explanation
The p-value is the probability of observing your data (or more 
extreme) IF the Null Hypothesis were actually true. A small p-value 
means the observed result would be unlikely under H0, giving 
evidence to reject it. The Significance Level (α, commonly 0.05) 
is the threshold we compare the p-value against BEFORE running the test.

### Formula
Decision rule: if p-value < α → reject H0 (result is statistically significant)
if p-value ≥ α → fail to reject H0

### Business Example
If α = 0.05 and a website redesign test gives p-value = 0.01, we 
reject H0 - the conversion rate difference is statistically 
significant, not just random noise.

### AI/ML Example
When comparing model performance, a p-value below the significance 
level confirms one model is genuinely better, not just better by 
random luck on this particular test set.

```python
alpha = 0.05
p_value = 0.01   # from our earlier test

if p_value < alpha:
    print(f"p-value ({p_value}) < alpha ({alpha}) -> Reject H0. Result is statistically significant.")
else:
    print(f"p-value ({p_value}) >= alpha ({alpha}) -> Fail to reject H0. Not statistically significant.")

# Where it's used: comparing p_value directly against alpha
# implements the standard hypothesis testing decision rule - here,
# since 0.01 < 0.05, we'd conclude the new design genuinely
# performs differently, not just by chance.
```

## 3. Confidence Interval

### Explanation
A Confidence Interval gives a RANGE of values likely to contain 
the true population parameter, with a stated confidence level 
(commonly 95%) - instead of a single point estimate, it 
communicates the uncertainty around that estimate.

### Formula
CI = sample mean ± (critical value × standard error)

### Business Example
A survey estimates average customer satisfaction as 7.5/10, with a 
95% Confidence Interval of [7.2, 7.8] - meaning we're 95% confident 
the TRUE average (across all customers) falls in that range.

### AI/ML Example
Confidence Intervals are used to report model performance metrics 
with uncertainty (e.g., "accuracy: 92% ± 2%"), giving a more honest 
picture than a single number alone.

```python
import numpy as np
from scipy import stats

data = [7.2, 7.5, 7.8, 7.1, 7.6, 7.4, 7.9, 7.3]
mean = np.mean(data)
sem = stats.sem(data)   # standard error of the mean
confidence_interval = stats.t.interval(0.95, len(data)-1, loc=mean, scale=sem)

print("Sample Mean:", mean)
print("95% Confidence Interval:", confidence_interval)

# Where it's used: stats.t.interval() calculates the range within
# which we're 95% confident the TRUE population mean lies, based on
# this sample - wider intervals mean more uncertainty, narrower
# means more precision.
```

## 4. Type I Error and Type II Error

### Explanation
Type I Error (False Positive): rejecting H0 when it was actually 
TRUE - concluding there's an effect when there isn't one. Type II 
Error (False Negative): failing to reject H0 when it was actually 
FALSE - missing a real effect that exists.

### Business Example
Type I: Concluding a new drug works when it actually doesn't 
(dangerous - patients get a useless treatment). Type II: 
Concluding a new drug doesn't work when it actually does 
(missed opportunity - a genuinely helpful drug gets rejected).

### AI/ML Example
In fraud detection: Type I Error = flagging a legitimate 
transaction as fraud (annoys customers). Type II Error = missing 
an actual fraudulent transaction (real financial loss) - the 
tradeoff between these two errors shapes how a fraud model's 
threshold is set.

```python
# Simulating the tradeoff conceptually
true_situation = "H0 is actually FALSE (drug really works)"
our_decision = "We failed to reject H0 (concluded drug doesn't work)"

print("Type II Error scenario:")
print("Reality:", true_situation)
print("Our conclusion:", our_decision)
print("Result: We MISSED a real effect - a Type II Error (False Negative)")

# Where it's used: this simply illustrates the LOGIC of Type II
# Error conceptually - the actual detection of these errors happens
# through repeated testing/power analysis, which is a more advanced
# topic beyond this notebook's scope.
```

## 5. One-tailed Test and Two-tailed Test

### Explanation
A One-tailed Test checks for an effect in ONE specific direction 
only (e.g., "is the new method BETTER" - not caring if it's worse). 
A Two-tailed Test checks for a difference in EITHER direction 
(e.g., "is the new method DIFFERENT" - better OR worse both count).

### Business Example
One-tailed: Testing if a new ad campaign INCREASES sales (only 
care about improvement). Two-tailed: Testing if a new supplier's 
product quality is DIFFERENT from the current one (could be better 
or worse - both matter).

### AI/ML Example
One-tailed tests are used when you specifically want to prove a 
new model is BETTER than the current one. Two-tailed tests are 
used when you're checking if two models perform DIFFERENTLY at 
all, regardless of direction.

```python
from scipy import stats

group_a = [50, 55, 52, 58, 54]
group_b = [60, 62, 65, 59, 63]

# Two-tailed test (default) - checks for ANY difference
t_stat, p_two_tailed = stats.ttest_ind(group_a, group_b)
print("Two-tailed p-value:", p_two_tailed)

# One-tailed test - checking specifically if group_b > group_a
p_one_tailed = p_two_tailed / 2   # simple conversion when the direction matches what we expect
print("One-tailed p-value (testing if B > A):", p_one_tailed)

# Where it's used: the two-tailed p-value tests for ANY difference
# in either direction. Dividing by 2 for the one-tailed version
# (valid when the observed direction matches our specific
# hypothesis) makes the test MORE sensitive to detecting an effect
# in that one specific direction we care about.
```
