Statistics & Mathematics for AI/ML Engineers

This notebook explains how statistical concepts covered earlier in 
this sprint are directly applied throughout a typical ML workflow.

## 1. Data Cleaning
Statistics helps identify inconsistencies - e.g., checking value 
ranges, frequency counts, and summary stats (mean/min/max) to spot 
impossible or erroneous values (like a negative age) before they 
corrupt a model.

## 2. Missing Value Handling
Mean/Median/Mode (Notebook 2) are directly used to IMPUTE missing 
values - median preferred for skewed data, mode for categorical data.

## 3. Outlier Detection
IQR and Z-Score methods (Notebook 9) are used to detect and 
handle extreme values that could distort model training.

## 4. Feature Engineering
Descriptive statistics guide creation of new features - e.g., 
creating a "distance from mean" feature, or binning continuous data 
based on quartiles.

## 5. Feature Scaling
Standard Deviation and Mean (Notebook 3) are the basis of 
Standardization (Z-score scaling) - transforming features to have 
mean=0, std=1.

## 6. Data Normalization
Range (Notebook 3) is used in Min-Max Scaling, rescaling values to 
a fixed range like [0,1] using the minimum and maximum.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

# Standardization (uses mean, std)
standardized = (data - np.mean(data)) / np.std(data)
print("Standardized:", standardized)

# Normalization (uses min, max / range)
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print("Normalized:", normalized)

# Where it's used: standardized uses MEAN and STD (Notebook 3
# concepts) directly. normalized uses MIN/MAX (Range concept) -
# both are direct statistical applications in preprocessing.
```

## 7. Model Evaluation
Statistical concepts like Mean Squared Error (uses Variance/squared 
deviations, Notebook 3), Confidence Intervals (Notebook 10), and 
hypothesis testing (comparing two models' performance) are all used 
to evaluate and compare models rigorously.

## 8. Linear Regression
Built directly on statistical concepts - minimizes squared errors 
(variance-based), assumes normally distributed residuals, and uses 
correlation (Notebook 8) to understand feature-target relationships.

## 9. Logistic Regression
Outputs a PROBABILITY (Notebook 5 concepts), uses Maximum Likelihood 
Estimation (a statistical estimation method) to fit the model.

## 10. Naive Bayes
Directly implements Bayes' Theorem (Notebook 5) - calculates 
P(class | features) using prior probabilities and likelihoods from 
training data.

## 11. Clustering
Uses distance/variance concepts - algorithms like K-Means minimize 
within-cluster variance (Notebook 3) to group similar data points together.

## 12. Principal Component Analysis (PCA)
Directly built on Variance and Covariance (Notebooks 3 & 8) - finds 
directions (principal components) that capture the MOST variance in 
the data.

## 13. Recommendation Systems
Use Correlation (Notebook 8) to find similar users/items (e.g., 
users who rate similarly), and probability to estimate likelihood 
of a user liking an item.

```python
import numpy as np
from sklearn.naive_bayes import GaussianNB

# Simple demonstration: Naive Bayes directly uses Bayes' Theorem
X = np.array([[1, 2], [2, 3], [3, 3], [6, 8], [7, 8], [8, 9]])
y = np.array([0, 0, 0, 1, 1, 1])   # two classes

model = GaussianNB()
model.fit(X, y)
prediction = model.predict([[2, 2]])
probabilities = model.predict_proba([[2, 2]])

print("Predicted class:", prediction)
print("Class probabilities:", probabilities)

# Where it's used: GaussianNB() internally applies Bayes' Theorem -
# calculating P(class | features) using the mean/std (Gaussian 
# assumption) of each feature per class, learned from training data.
```
