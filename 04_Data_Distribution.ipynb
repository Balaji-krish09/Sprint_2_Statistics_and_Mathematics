## 1. Normal Distribution

### Definition
The Normal Distribution is a symmetric, bell-shaped distribution 
where most values cluster around the mean, and values become less 
frequent the further they are from the mean in either direction. 
It's one of the most common distributions in nature and statistics.

### Formula
f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))
Where μ = mean, σ = standard deviation

### Real-World Example
Human height across a large population tends to follow a normal 
distribution - most people cluster around average height, with 
fewer people being extremely short or extremely tall.

### AI/ML Example
Many ML algorithms (like Linear Regression) ASSUME errors/residuals 
are normally distributed. Feature scaling techniques like 
Standardization transform data to approximate a normal distribution 
with mean 0, std 1.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=1000)   # mean=50, std=10

plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(data), "| Std Dev:", np.std(data))

# Where it's used: np.random.normal(loc=50, scale=10, size=1000)
# generates 1000 values following a normal distribution centered at
# 50. The histogram visually confirms the classic bell shape - most
# values cluster near 50, tapering off symmetrically on both sides.
```

## 2. Uniform Distribution

### Definition
The Uniform Distribution is one where EVERY value within a given 
range has an EQUAL chance of occurring - no clustering around any 
particular value, unlike the Normal Distribution.

### Formula
f(x) = 1 / (b - a), for a ≤ x ≤ b (where a=min, b=max)

### Real-World Example
Rolling a fair six-sided die - each number (1 through 6) has 
exactly the same 1/6 probability of appearing, no number is more 
"likely" than another.

### AI/ML Example
Uniform Distribution is used to initialize weights in some neural 
networks, and it's the default distribution assumed by many random 
number generators used in train/test splitting or shuffling data.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(low=0, high=100, size=1000)

plt.hist(data, bins=30, color="lightgreen", edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# Where it's used: np.random.uniform(low=0, high=100, size=1000)
# generates values spread EVENLY across the 0-100 range. Notice the
# histogram bars are roughly the SAME height across the whole
# range - unlike Normal Distribution's bell shape, there's no
# clustering around any particular value here.
```

## 3. Gaussian Distribution

### Definition
"Gaussian Distribution" is just ANOTHER NAME for the Normal 
Distribution, named after mathematician Carl Friedrich Gauss. 
There's no mathematical difference - it's the same bell-shaped 
distribution we just covered.

### Formula
Same as Normal Distribution: f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

### Real-World Example
Measurement errors in scientific experiments (like measuring a 
chemical's weight repeatedly) typically follow a Gaussian 
distribution - small errors are common, large errors are rare.

### AI/ML Example
"Gaussian" appears throughout ML terminology - Gaussian Naive 
Bayes (a classification algorithm), Gaussian Mixture Models 
(clustering), and Gaussian noise (added intentionally in some 
data augmentation techniques) - all referring to this same 
underlying distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Identical code to Normal Distribution - because it IS the same thing
data = np.random.normal(loc=0, scale=1, size=1000)   # "standard" Gaussian: mean=0, std=1

plt.hist(data, bins=30, color="salmon", edgecolor="black")
plt.title("Gaussian (Normal) Distribution - Standard Form")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# Where it's used: this code is IDENTICAL in structure to the
# Normal Distribution code - confirming Gaussian and Normal are the
# same concept. Here we used mean=0, std=1, called the "Standard
# Normal" or "Standard Gaussian" distribution, commonly used as a
# reference/default in statistics.
```

## 4. Left Skew and Right Skew

### Definition
Skewness measures the ASYMMETRY of a distribution. Right Skew 
(positive skew) has a longer tail on the RIGHT side - most values 
cluster on the lower end, with a few large outliers pulling the 
tail right. Left Skew (negative skew) is the opposite - a longer 
tail on the LEFT, most values cluster on the higher end.

### Formula
Skewness = [Σ(x - mean)³ / n] / (Standard Deviation)³

### Real-World Example
Income distribution is typically RIGHT-skewed - most people earn 
moderate incomes, but a few very high earners create a long tail 
stretching to the right. Age at retirement tends to be LEFT-skewed - 
most people retire around a typical age, with a shorter tail of 
people retiring very early.

### AI/ML Example
Detecting skewness is a key data preprocessing step - highly skewed 
features (like income or transaction amounts) are often log-
transformed before training a model, since many algorithms perform 
better with more symmetric feature distributions.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

right_skewed = np.random.exponential(scale=2, size=1000)   # naturally right-skewed
left_skewed = -np.random.exponential(scale=2, size=1000)   # flipped to be left-skewed

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(right_skewed, bins=30, color="orange", edgecolor="black")
axes[0].set_title("Right Skew (Positive)")
axes[1].hist(left_skewed, bins=30, color="purple", edgecolor="black")
axes[1].set_title("Left Skew (Negative)")
plt.show()

print("Right skew value:", skew(right_skewed))   # positive number
print("Left skew value:", skew(left_skewed))     # negative number

# Where it's used: skew() from scipy.stats calculates the actual
# skewness NUMBER. A positive value confirms right skew (long tail
# right), a negative value confirms left skew (long tail left) -
# matching what we can visually see in the two histograms.
```

## 5. Symmetric Distribution

### Definition
A Symmetric Distribution is one where the LEFT and RIGHT sides are 
mirror images of each other around the center - skewness is 
approximately zero. The Normal/Gaussian Distribution is the 
classic example.

### Formula
Symmetric when: Skewness ≈ 0

### Real-World Example
Test scores on a well-designed standardized exam often approximate 
a symmetric distribution - roughly equal numbers of students score 
above and below the average, tapering off similarly on both sides.

### AI/ML Example
Many statistical tests (like t-tests) assume the underlying data is 
approximately symmetric/normal - checking for symmetry is a common 
early step before applying such tests to real data.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

symmetric_data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(symmetric_data, bins=30, color="lightblue", edgecolor="black")
plt.title("Symmetric Distribution")
plt.axvline(np.mean(symmetric_data), color="red", linestyle="--", label="Mean")
plt.legend()
plt.show()

print("Skewness (should be close to 0):", skew(symmetric_data))

# Where it's used: the histogram visually shows roughly equal
# spread on both sides of the mean (red dashed line). The skew()
# value confirms this numerically - it should be very close to 0,
# unlike the clearly positive/negative values we saw for
# right/left-skewed data earlier.
```

## 6. Kurtosis

### Definition
Kurtosis measures how HEAVY or LIGHT the TAILS of a distribution 
are compared to a normal distribution - it tells you how likely 
extreme values (outliers) are. High kurtosis = heavier tails, more 
extreme outliers likely. Low kurtosis = lighter tails, fewer extremes.

### Formula
Kurtosis = [Σ(x - mean)⁴ / n] / (Standard Deviation)⁴ - 3
(the "-3" makes Normal Distribution's kurtosis = 0, called "excess kurtosis")

### Real-World Example
Stock market returns often have HIGH kurtosis ("fat tails") - 
extreme market crashes or spikes happen more often than a normal 
distribution would predict, which is a well-known risk in finance.

### AI/ML Example
Kurtosis is checked during data preprocessing to identify features 
prone to extreme outliers - high-kurtosis features may need robust 
scaling techniques instead of standard scaling, which is more 
sensitive to extreme values.

```python
from scipy.stats import kurtosis
import numpy as np

normal_data = np.random.normal(0, 1, 1000)
heavy_tailed_data = np.random.standard_t(df=3, size=1000)   # t-distribution has heavier tails

print("Kurtosis of Normal data (~0 expected):", kurtosis(normal_data))
print("Kurtosis of heavy-tailed data (higher, positive):", kurtosis(heavy_tailed_data))

# Where it's used: kurtosis() from scipy.stats measures tail
# heaviness. The heavy_tailed_data (generated from a t-distribution,
# known for extreme values) shows noticeably HIGHER kurtosis than
# normal_data - confirming it's more prone to extreme outliers.
```

## 7. Empirical Rule (68-95-99.7)

### Definition
The Empirical Rule applies specifically to Normal Distributions: 
approximately 68% of data falls within 1 Standard Deviation of the 
mean, 95% within 2 Standard Deviations, and 99.7% within 3 
Standard Deviations.

### Formula
68% → mean ± 1σ
95% → mean ± 2σ
99.7% → mean ± 3σ

### Real-World Example
If adult heights have a mean of 170cm and SD of 10cm, about 68% of 
people are between 160-180cm, 95% are between 150-190cm, and 99.7% 
are between 140-200cm.

### AI/ML Example
The Empirical Rule is the basis for a common outlier detection 
method: flagging any data point beyond 3 standard deviations from 
the mean as a likely outlier (assuming roughly normal data), used 
frequently in anomaly detection systems.

```python
import numpy as np

data = np.random.normal(loc=170, scale=10, size=10000)
mean = np.mean(data)
std = np.std(data)

within_1sd = np.sum((data > mean - std) & (data < mean + std)) / len(data) * 100
within_2sd = np.sum((data > mean - 2*std) & (data < mean + 2*std)) / len(data) * 100
within_3sd = np.sum((data > mean - 3*std) & (data < mean + 3*std)) / len(data) * 100

print(f"Within 1 SD: {within_1sd:.1f}% (expected ~68%)")
print(f"Within 2 SD: {within_2sd:.1f}% (expected ~95%)")
print(f"Within 3 SD: {within_3sd:.1f}% (expected ~99.7%)")

# Where it's used: this code counts what percentage of the actual
# generated data falls within 1, 2, and 3 standard deviations of
# the mean - and confirms it closely matches the theoretical
# 68-95-99.7 rule, since our data was generated from a true normal distribution.
```
