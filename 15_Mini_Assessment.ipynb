### Coding Exercise 1: Write Python code to calculate Variance.

```python
import numpy as np

data = [10, 20, 30, 40, 50]
variance = np.var(data)
print("Variance:", variance)
```

### Coding Exercise 2: Detect Outliers using the IQR Method.

```python
import numpy as np

data = np.array([10, 12, 14, 15, 16, 18, 100])
q1, q3 = np.percentile(data, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr

outliers = data[(data < lower) | (data > upper)]
print("Outliers:", outliers)
```

### Coding Exercise 3: Calculate Correlation between two variables.

```python
import numpy as np

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation = np.corrcoef(x, y)[0][1]
print("Correlation:", correlation)
```
